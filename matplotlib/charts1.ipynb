```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Barplot

```python
dept = ["CSE", "AIDS", "AIML", "ECE", "MECH", "CIVIL"]
enroll = [95, 75, 80, 65, 81, 60]

stucount = {
    "Dept": dept,
    "Enrolled": enroll
}

student_df = pd.DataFrame(stucount)


plt.figure(figsize=(9, 5))

bar=plt.bar(
    student_df["Dept"],
    student_df["Enrolled"],
    color=["#CA5B64", "#8DE588", "#F2A2A1", "#DBB8D1", "#86FA2E", "#EAEA3D"],edgecolor="white",width=0.6

)
#title and label
plt.title("Dept wise student count",fontsize=15,fontweight='bold',pad=15)
plt.xlabel("name of dept",fontsize=12)
plt.ylabel("no. of student",fontsize=12)
#advance
plt.bar_label(bar, fontsize=12, padding=5, fontweight='bold')

plt.grid(axis='y', linestyle=":", alpha=0.45)
plt.ylim(0, 110)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

lineplot

```python
weekly_sales={
    "Days":['Mon','tue','wed','thu','fri','sat','sun'],
    "week1_sales":[32000,28000,42000,35000,37000,32000,40000],
    "week2_sales":[35000,30000,40000,35000,34000,42000,38000],
}
sales=pd.DataFrame(weekly_sales)
#----------------week 1 sales------------------
plt.plot(sales["Days"],sales["week1_sales"],color='red',ls='-',marker='d',label="week1") #o,*,d

#----------------week 2 sales------------------
plt.plot(sales["Days"],sales["week2_sales"],color='green',ls='--',marker='d',label="week2 ") #o,*,d
#create title and label
plt.title("weekly sales overview",fontsize=15,fontweight='bold',pad=15)
plt.xlabel("name of days",fontsize=12)
plt.ylabel("total amnt of sales",fontsize=12)
#advanced
plt.legend()
plt.ylim(25000,50000)
plt.tight_layout()
plt.show()
```

scatter plot

```python
study_hours = np.random.randint(3, 10, 50)
marks = np.random.randint(300, 500, 50)

studentdata = {
    "Study_hours": study_hours,
    "Marks": marks
}

student_df = pd.DataFrame(studentdata)

plt.scatter(student_df["Marks"], student_df["Study_hours"],color="blue",alpha=0.5,
s=200,label="marks")
plt.xlabel("Marks")
plt.ylabel("Study Hours")
plt.title("Study Hours vs Marks")

plt.show()
```

Pie Chart

```python
weekly_sales={
    "brand":["oneplus","apple","oppo","vivo","samsung"],
    "unit" :[23,12,25,28,19]
}
salesdf=pd.DataFrame(weekly_sales)

plt.pie(salesdf['unit'],
labels=salesdf["brand"],
autopct="%.2f%%",
explode=[0,0,0,0,0.1], shadow="True",startangle=90
)
plt.title("weekly sales overview",fontsize=12, fontweight='bold')
plt.legend(
    loc='right',bbox_to_anchor=(1.5,1)
)
plt.tight_layout()
plt.show()
```

Stack plot

```python

import pandas as pd
import matplotlib.pyplot as plt

weekly_sales = {
    "Days": ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"],
    "Week1": [5, 10, 30, 20, 35, 60, 80],
    "Week2": [30, 60, 30, 70, 80, 50, 30],
    "Week3": [10, 40, 50, 23, 69, 82, 43]
}

sales = pd.DataFrame(weekly_sales)

plt.stackplot(
    sales["Days"],
    sales["Week1"],
    sales["Week2"],
    sales["Week3"],
    labels=["Week1", "Week2", "Week3"]
)

plt.grid(axis="both", linestyle="--", alpha=0.9)
plt.legend()
plt.show()
```

boxplot

```python

import pandas as pd
import matplotlib.pyplot as plt

weekly_sales = {
    "Days": ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"],
    "Week1": [5, 10, 30, 20, 35, 60, 80],
    "Week2": [30, 60, 30, 70, 80, 50, 30],
    "Week3": [10, 40, 50, 23, 69, 82, 43]
}

sales = pd.DataFrame(weekly_sales)

plt.boxplot(
    [sales["Week1"], sales["Week2"], sales["Week3"]],
    tick_labels=["Week1", "Week2", "Week3"]
)

plt.ylabel("Sales (units)")
plt.title("Weekly Sales Box Plot")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

stemplot

```python
plt.stem(sales["Days"], sales["Week1"], label="Week1")
plt.stem(sales["Days"], sales["Week2"], label="Week2")
plt.stem(sales["Days"], sales["Week3"], label="Week3")

plt.xlabel("Days")
plt.ylabel("Sales (units)")
plt.title("Weekly Sales Stem Plot")
plt.legend()
plt.grid(axis="both", linestyle="--", alpha=0.5)
plt.show()
```

step plot

```python
plt.step(sales["Days"], sales["Week1"], label="Week1", where="mid")
plt.step(sales["Days"], sales["Week2"], label="Week2", where="mid")
plt.step(sales["Days"], sales["Week3"], label="Week3", where="mid")

plt.xlabel("Days")
plt.ylabel("Sales (units)")
plt.title("Weekly Sales Step Plot")
plt.legend()
plt.grid(axis="both", linestyle="--", alpha=0.5)
plt.show()
```
