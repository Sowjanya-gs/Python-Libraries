```python
import pandas as pd
```

```python
exp=pd.read_excel("dataset_pandas/expense.xlsx")
exp
```

```python
exp.groupby('Category')['Amount'].sum()
exp.groupby('Category')['Amount'].agg(['sum'])
exp.groupby('Category')['Amount'].agg(['sum','count','max','min'])
exp.groupby('Payment Mode')['Amount'].agg(['sum','count'])
exp.groupby(['Payment Mode','Category'])['Amount'].agg(['sum'])
exp.groupby('Category')['Amount'].sum().sort_values(ascending=False)
```

```python
#---------------Task-----------------------
# Create a DataFrame containing the total expense of each category.
# What is the average transaction amount for each payment mode?
# Find the highest individual transaction in each category
# Show total, average, minimum and maximum transaction amount for each payment mode.

total_expense = exp.groupby('Category')['Amount'].sum().reset_index()

print(total_expense)



average_transaction = exp.groupby('Payment Mode')['Amount'].mean().reset_index()

print(average_transaction)



highest_transaction = exp.groupby('Category')['Amount'].max().reset_index()

print(highest_transaction)



payment_summary = exp.groupby('Payment Mode')['Amount'].agg(
    Total='sum',
    Average='mean',
    Minimum='min',
    Maximum='max'
).reset_index()

print(payment_summary)
```

Merge or join

```python
customer=pd.read_csv('dataset_pandas/customer.csv')
customer
payment=pd.read_csv('dataset_pandas/payment.csv')
payment
```

```python
#join
#innerjoin
pd.merge(left=customer,right=payment,on="customerid",how="inner")
#left join
pd.merge(left=customer,right=payment,on="customerid",how="left")
#right join
pd.merge(left=customer,right=payment,on="customerid",how="right")
#cross join task
pd.merge(left=customer, right=payment, how="cross")
```
