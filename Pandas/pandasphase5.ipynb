Compare,Pivoting and Melting

```python
import pandas as pd
```

compare

```python
shop_oct8={
    "Fruits":['mango','apple','banana','papaya'],
    "price":[100,150,60,30],
    "qunt":[10,8,12,15]
}
stock_oct8=pd.DataFrame(shop_oct8)
shop_oct9={
    "Fruits":['mango','apple','banana','papaya'],
    "price":[120,200,60,35],
    "qunt":[15,6,24,13]
}
stock_oct9=pd.DataFrame(shop_oct9)
#-------------------compare-----------------------
stock_oct8.compare(stock_oct9)
stock_oct8.compare(stock_oct9,align_axis=0)
stock_oct8.compare(stock_oct9,align_axis=0,keep_equal=True)
stock_oct8.compare(stock_oct9,align_axis=0,keep_equal=True,keep_shape=True)
```

pivoting

```python
monthly_sales={
    "products":['laptop','mobile','laptop','mobile','laptop','laptop'],
    "month":['jan','jan','feb','feb','mar','apr'],
     "price":[65000,25000,78500,17500,75000,45000]
}
sales=pd.DataFrame(monthly_sales)
#----------pivoting--------------
sales.pivot(
            columns="month",
            index="products",
            values="price")
```

```python
monthly_sales={
    "products":['laptop','mobile','laptop','mobile','laptop','laptop'],
    "month":['jan','jan','feb','feb','mar','mar'],
     "price":[65000,25000,78500,17500,75000,45000]
}
sales=pd.DataFrame(monthly_sales)
#----------pivoting--------------
sales.pivot_table(
            columns="month",
            index="products",
            values="price",
            aggfunc="sum")
```

melting

```python
shop_sales={
    "products":['laptop','mobile','mouse'],
    "jan":[50000,15000,150],
    "feb":[65000,25000,340],
    "march":[75000,13000,560]

}
sales=pd.DataFrame(shop_sales)
#-------melting--------
sales.melt(id_vars="products",
           var_name="month",
           value_name="sales")
```
